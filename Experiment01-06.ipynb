## 5. Variable Classification

The variables in the dataset can be classified as follows:

| Variable | Type | Description |
|---|---|---|
| Pregnancies | Numerical | Number of pregnancies |
| Glucose | Numerical | Plasma glucose concentration |
| BloodPressure | Numerical | Diastolic blood pressure |
| SkinThickness | Numerical | Triceps skin fold thickness |
| Insulin | Numerical | 2-hour serum insulin |
| BMI | Numerical | Body Mass Index |
| DiabetesPedigreeFunction | Numerical | Diabetes pedigree function |
| Age | Numerical | Age of the patient |
| Outcome | Binary | Diabetes outcome (0 = No, 1 = Yes) |

```python
print("Unique values in Outcome:")
print(df['Outcome'].unique())

print("\nFrequency of each outcome:")
print(df['Outcome'].value_counts())
```

## 7. Missing Value Analysis

The dataset is examined for missing values using the `isnull()` function.

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:")
print(df.isnull().sum().sum())
```

### Investigation of Zero Values

Zero values in Glucose, BloodPressure, SkinThickness, Insulin, and BMI are investigated because they may represent missing or invalid measurements rather than genuine medical values.

```python
medical_columns = [
    'Glucose',
    'BloodPressure',
    'SkinThickness',
    'Insulin',
    'BMI'
]
zero_counts = (df[medical_columns] == 0).sum()
print("Number of zero values:")
print(zero_counts)
```

```python
zero_percentage = (df[medical_columns] == 0).mean() * 100
print("Percentage of zero values:")
print(zero_percentage.round(2))
```

## 9. Outlier Analysis


```python
plt.figure(figsize=(14, 6))
sns.boxplot(data=df)
plt.title("Boxplot of Numerical Variables")
plt.xticks(rotation=45)
plt.show()
```

```python
Q1 = df.quantile(0.25, numeric_only=True)
Q3 = df.quantile(0.75, numeric_only=True)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outlier_counts = ((df < lower_bound) | (df > upper_bound)).sum()
print("Potential outliers in each numerical variable:")
print(outlier_counts)
```

## 10. Data Visualization


```python
plt.figure(figsize=(6, 4))
sns.countplot(x='Outcome', data=df)
plt.title("Frequency of Diabetes Outcomes")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Number of Patients")
plt.show()
```

```python
df.hist(figsize=(14, 10), bins=20)
plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x='Glucose',
    y='BMI',
    hue='Outcome'
)
plt.title("Glucose vs BMI by Diabetes Outcome")
plt.xlabel("Glucose")
plt.ylabel("BMI")
plt.show()
```
