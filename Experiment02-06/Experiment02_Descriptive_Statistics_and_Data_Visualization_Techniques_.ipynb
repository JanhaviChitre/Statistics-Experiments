## Experiment 02
Descriptive Statistics and Data Visualization
Pima Indians Diabetes Dataset

## Aim
To apply descriptive statistical measures and data visualization techniques to summarize, analyze, and interpret the distribution and relationships present in the Pima Indians Diabetes Dataset.

## Objectives
1. To apply descriptive statistical and visualization techniques to explore the characteristics of a real-world healthcare dataset.
2. To interpret distributions, variability, relationships, and patterns obtained from statistical analysis and visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
from google.colab import files
uploaded = files.upload()
df = pd.read_csv("diabetes.csv")
print("Dataset loaded successfully.")
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
print("\nFirst 5 Records:")
display(df.head())
```

```python
print("Descriptive Statistics:")
display(df.describe())
```

```python
print("Mean:")
display(df.mean(numeric_only=True))
```

```python
print("Median:")
display(df.median(numeric_only=True))
```

```python
print("Mode:")
display(df.mode(numeric_only=True).iloc[0])
```

```python
numeric_df = df.select_dtypes(include=np.number)
minimum = numeric_df.min()
maximum = numeric_df.max()
data_range = maximum - minimum
range_summary = pd.DataFrame({
    "Minimum": minimum,
    "Maximum": maximum,
    "Range": data_range
})
display(range_summary)
```

```python
variance = numeric_df.var()
std_deviation = numeric_df.std()
variability = pd.DataFrame({
    "Variance": variance,
    "Standard Deviation": std_deviation
})
display(variability)
```

```python
numeric_features = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]
df[numeric_features].hist(
    figsize=(14, 10),
    bins=20
)
plt.suptitle("Distribution of Numerical Variables", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 7))
sns.boxplot(data=df[numeric_features])
plt.title("Boxplots of Numerical Variables")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))
sns.countplot(data=df, x="Outcome")
plt.title("Diabetes Outcome Frequency")
plt.xlabel("Outcome (0 = Non-diabetic, 1 = Diabetic)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="Glucose",
    y="BMI",
    hue="Outcome"
)
plt.title("Glucose vs BMI")
plt.xlabel("Glucose")
plt.ylabel("BMI")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df,
    x="Age",
    y="Glucose",
    hue="Outcome"
)
plt.title("Age vs Glucose")
plt.xlabel("Age")
plt.ylabel("Glucose")
plt.show()
```

```python
pairplot_data = df[
    ["Glucose", "BMI", "Age", "BloodPressure", "Outcome"]
]
sns.pairplot(
    pairplot_data,
    hue="Outcome"
)
plt.show()
```

## Conclusion
Descriptive statistical measures and visualization techniques were applied to the Pima Indians Diabetes Dataset to understand the distribution, variability, and relationships among the variables.
